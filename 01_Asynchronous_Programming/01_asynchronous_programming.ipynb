## Synchronous Programming

```python
import time
```

```python
def fetch_weather():
    print("Fetching weather data...")
    time.sleep(4)  # Simulate a delay in fetching data
    print("Weather data fetched.")


def fetch_news():
    print("Fetching news data...")
    time.sleep(2)  # Simulate a delay in fetching data
    print("News data fetched.")


#these both function are indepent function 

def main():
    start_time = time.time()

    fetch_weather()
    fetch_news()

    end_time = time.time()
    print(f"Total time taken: {end_time - start_time:.2f} seconds")

#but when we run our main code first function will run and after that
#  second function will run so it will take 6 seconds to run both function 
#but there are independent to each other  


main()
```

## Asynchronous Programming

```python
import asyncio

async def fetch_weather():
    print("Fetching weather data...")
    await asyncio.sleep(4)
    #we are writing await here becuse we are using async function
    #  so we have to use await to wait for the function to complete  
    # so by writing await here when our python interpreter will come 
    # here by await we are telling it will take time here 
    # so it will not wait for this function(execution) to complete it 
    # will go to next function(execution/code it have)

    #so when this code is executing at the same time our next function
    #  will also execute so it will take
    print("Weather data fetched.")


async def fetch_news():
    print("Fetching news data...")
    await asyncio.sleep(2)  # Simulate a delay in fetching data
    print("News data fetched.")


#these both function are indepent function 

async def main():
    start_time = time.time()

    await asyncio.gather(fetch_weather(), fetch_news()) #used to run multiple asynchronous operations concurrently

    end_time = time.time()
    print(f"Total time taken: {end_time - start_time:.2f} seconds")

#but when we run our main code first function will run and after that
#  second function will run so it will take 6 seconds to run both function 
#but there are independent to each other  


await main()
```
